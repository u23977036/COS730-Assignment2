```python
class Validator:
    def validate_format(self, artefact_data):
        return artefact_data.endswith(".pdf")


class Database:
    def save_submission(self, artefact_data):
        return "confirmation"

    def fetch_reviewers(self):
        return ["Reviewer 1", "Reviewer 2", "Reviewer 3"]


class ReviewerManager:
    def __init__(self, database):
        self.database = database

    def get_available_reviewers(self):
        return self.database.fetch_reviewers()

    def start_conflict_check(self, reviewers):
        return reviewers

    def check_workload(self, reviewers):
        return reviewers

    def share_reviewers(self, reviewers):
        return reviewers


class Reviewer:
    def assign_review(self, reviewer):
        pass

    def submit_score(self, reviewer):
        return 75


class EvaluationManager:
    def start_evaluation(self):
        pass

    def collect_scores(self, scores):
        return scores

    def calculate_average(self, scores):
        return sum(scores) / len(scores)

    def check_consensus(self, scores):
        return max(scores) - min(scores) <= 20

    def apply_rules(self, average, consensus):
        if average >= 70 and consensus:
            return "accepted"
        elif average < 50:
            return "rejected"
        else:
            return "revision"


class NotificationService:
    def notify_acceptance(self):
        return "Submission accepted."

    def notify_rejection(self):
        return "Submission rejected."

    def notify_revision(self):
        return "Submission requires revision."


class SubmissionController:
    def __init__(
        self,
        validator,
        database,
        reviewer_manager,
        reviewer,
        evaluation_manager,
        notification_service
    ):
        self.validator = validator
        self.database = database
        self.reviewer_manager = reviewer_manager
        self.reviewer = reviewer
        self.evaluation_manager = evaluation_manager
        self.notification_service = notification_service

    def submit(self, artefact_data):
        print("=== Baseline Implementation ===\n")

        print("UI: submitResearchOutput(data)")
        print("SubmissionController: submit(data)\n")

        print("Validator: validateFormat(data)")
        valid = self.validator.validate_format(artefact_data)

        if not valid:
            print("Validation failed.\n")
            print("=== Final Result ===")
            print("Error: Invalid submission format.")
            return "Error: Invalid submission format."

        print("Validation successful.\n")

        print("Database: saveSubmission(data)")
        confirmation = self.database.save_submission(artefact_data)
        print(f"Database returned: {confirmation}\n")

        print("ReviewerManager: getAvailableReviewers()")
        reviewers = self.reviewer_manager.get_available_reviewers()

        print("Database: fetchReviewers()\n")

        print("ReviewerManager: startConflictCheck(reviewersList)")
        reviewers = self.reviewer_manager.start_conflict_check(reviewers)

        print("ReviewerManager: checkWorkload(reviewersList)")
        reviewers = self.reviewer_manager.check_workload(reviewers)

        print("ReviewerManager: shareReviewers()")
        reviewers = self.reviewer_manager.share_reviewers(reviewers)
        print()

        for reviewer in reviewers:
            print(f"Reviewer: assignReview() assigned to {reviewer}")
            self.reviewer.assign_review(reviewer)

        print("\nEvaluationManager: startEvaluation()\n")
        self.evaluation_manager.start_evaluation()

        scores = []
        for reviewer in reviewers:
            print(f"Reviewer: submitScore(score) from {reviewer}")
            score = self.reviewer.submit_score(reviewer)
            scores.append(score)

        print("\nEvaluationManager: collectScores(score)")
        scores = self.evaluation_manager.collect_scores(scores)

        print("EvaluationManager: calculateAverage()")
        average = self.evaluation_manager.calculate_average(scores)

        print("EvaluationManager: checkConsensus()")
        consensus = self.evaluation_manager.check_consensus(scores)

        print("EvaluationManager: applyRules()")
        decision = self.evaluation_manager.apply_rules(average, consensus)
        print()

        if decision == "accepted":
            print("NotificationService: notifyAcceptance()\n")
            result = self.notification_service.notify_acceptance()
        elif decision == "rejected":
            print("NotificationService: notifyRejection()\n")
            result = self.notification_service.notify_rejection()
        else:
            print("NotificationService: notifyRevision()\n")
            result = self.notification_service.notify_revision()

        print("UI: sendNotification()\n")
        print("=== Final Result ===")
        print(result)

        return result


class UI:
    def __init__(self, submission_controller):
        self.submission_controller = submission_controller

    def submit_research_output(self, artefact_data):
        return self.submission_controller.submit(artefact_data)


# ---- MAIN EXECUTION ----

validator = Validator()
database = Database()
reviewer_manager = ReviewerManager(database)
reviewer = Reviewer()
evaluation_manager = EvaluationManager()
notification_service = NotificationService()

submission_controller = SubmissionController(
    validator,
    database,
    reviewer_manager,
    reviewer,
    evaluation_manager,
    notification_service
)

ui = UI(submission_controller)

# Run the system (DO NOT wrap in print)
result = ui.submit_research_output("research_output.pdf")
```
