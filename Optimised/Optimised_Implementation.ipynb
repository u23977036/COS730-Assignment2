```python
# Optimised Implementation
# Validator class
class Validator:

    # Check if the paper format is valid
    def validate(self, paper):

        print("Validator: validate(paper)")

        return True



# Database class
class Database:

    # Save submission into the database
    def save_submission(self, paper):

        print("Database: saveSubmission(paper)")

        return "SUB001"



# Reviewer manager class
class ReviewerManager:

    # Prepare and assign reviewers
    def prepare_reviewers(self, submission_id):

        print("ReviewerManager: prepareReviewers(submissionUniqueId)\n")

        reviewers = [
            "Reviewer 1",
            "Reviewer 2",
            "Reviewer 3"
        ]


        # Loop through reviewers
        for reviewer in reviewers:

            print(f"Checking conflict for {reviewer}")

            print(f"Checking workload for {reviewer}")

            print(f"Selecting suitable reviewer: {reviewer}")

            print(f"Sharing feedback meeting with {reviewer}\n")


        return reviewers


    # Submit paper for review
    def submit_for_review(self, submission_id, reviewers):

        print("ReviewerManager: submitForReview(submissionUniqueId, reviews)\n")


        reviews = [
            {"reviewer": "Reviewer 1", "score": 75},
            {"reviewer": "Reviewer 2", "score": 80},
            {"reviewer": "Reviewer 3", "score": 78}
        ]


        return reviews



# Evaluation manager class
class EvaluationManager:

    # Evaluate review scores
    def evaluate_reviews(self, submission_id, reviews):

        print("EvaluationManager: evaluateReviews(submissionUniqueId)\n")


        total = 0


        # Calculate total score
        for review in reviews:

            total += review["score"]


        average = total / len(reviews)


        # Assume reviewers reached consensus
        consensus = True


        return {
            "average": average,
            "consensus": consensus
        }



# Decision manager class
class DecisionManager:

    # Determine final decision
    def determine_decision(self, evaluation_result):

        print("DecisionManager: determineDecision(evaluationResult)\n")


        average = evaluation_result["average"]

        consensus = evaluation_result["consensus"]


        # Decision rules from decision table

        if average >= 70 and consensus:

            return "Submission accepted."


        elif average < 50:

            return "Submission rejected."


        else:

            return "Requires revision."



# Notification service class
class NotificationService:

    # Send notification to researcher
    def send_decision_notification(self, submission_id, decision):

        print("NotificationService: sendDecisionNotification(submissionUniqueId, decision)\n")

        print("Notification sent.\n")



# Main controller class
class SubmissionController:


    def __init__(
        self,
        validator,
        database,
        reviewer_manager,
        evaluation_manager,
        decision_manager,
        notification_service
    ):

        self.validator = validator

        self.database = database

        self.reviewer_manager = reviewer_manager

        self.evaluation_manager = evaluation_manager

        self.decision_manager = decision_manager

        self.notification_service = notification_service



    # Main system process
    def submit_paper(self, paper):

        print("=== Optimised System Execution ===\n")


        # Validate paper
        valid = self.validator.validate(paper)


        # Invalid submission
        if not valid:

            print("Validation failed.")

            return


        print("Validation successful.\n")


        # Save submission
        submission_id = self.database.save_submission(paper)

        print(f"Submission ID: {submission_id}\n")


        # Prepare reviewers
        reviewers = self.reviewer_manager.prepare_reviewers(
            submission_id
        )

        print("Reviewers assigned.\n")


        # Submit for review
        reviews = self.reviewer_manager.submit_for_review(
            submission_id,
            reviewers
        )

        print("Reviews submitted.\n")


        # Evaluate reviews
        evaluation_result = self.evaluation_manager.evaluate_reviews(
            submission_id,
            reviews
        )

        print(
            f"Average Score: {evaluation_result['average']}\n"
        )


        # Determine decision
        decision = self.decision_manager.determine_decision(
            evaluation_result
        )

        print(f"Decision: {decision}\n")


        # Send notification
        self.notification_service.send_decision_notification(
            submission_id,
            decision
        )


        print("=== Final Result ===")

        print(decision)



# Create objects

validator = Validator()

database = Database()

reviewer_manager = ReviewerManager()

evaluation_manager = EvaluationManager()

decision_manager = DecisionManager()

notification_service = NotificationService()



# Create controller

controller = SubmissionController(
    validator,
    database,
    reviewer_manager,
    evaluation_manager,
    decision_manager,
    notification_service
)



# Paper data

paper = {
    "title": "Optimised Research Submission System",
    "author": "Student"
}



# Run the system

controller.submit_paper(paper)
```
